## 29. Why Functions?

Suppose we repeatedly calculate GC content.

Without a function, we would repeatedly write the same calculation.

````python
sequences = ["ATGC", "GGCC", "ATAT", "GCGCGC"]

for sequence in sequences:
    gc = sequence.count("G") + sequence.count("C")
    percentage = gc / len(sequence) * 100
    print(sequence, percentage)
````

The loop is useful because it repeats the calculation.

But the calculation itself can be packaged into a reusable function.

## 30. Creating a Function

````python
def gc_content(dna):
    gc = dna.count("G") + dna.count("C")
    return gc / len(dna) * 100

print(gc_content("ATGC"))
print(gc_content("GGCC"))
print(gc_content("ATAT"))
````

Basic structure:

```python
def function_name(parameters):
    # function body
    return result
```

## 36. Functions + Loops

This is a very common bioinformatics pattern:

**store data → loop through data → call a function → report results**

````python
def gc_content(sequence):
    gc = sequence.count("G") + sequence.count("C")
    return (gc / len(sequence)) * 100

sequences = ["ATGC", "GGCC", "ATAT", "GCGCGC"]

for sequence in sequences:
    print(sequence, round(gc_content(sequence), 2))
````

## 37. Useful Bioinformatics Functions

````python
def sequence_length(sequence):
    return len(sequence)

def base_count(sequence, base):
    return sequence.count(base)

def gc_content(sequence):
    gc = sequence.count("G") + sequence.count("C")
    return (gc / len(sequence)) * 100

sequence = "ATGCGTACG"

print("Length:", sequence_length(sequence))
print("A:", base_count(sequence, "A"))
print("G:", base_count(sequence, "G"))
print("GC%:", round(gc_content(sequence), 2))
````

## 38. Reverse a Sequence

````python
def reverse_sequence(sequence):
    return sequence[::-1]

print(reverse_sequence("ATGCGT"))
````

## 39. DNA Complement

Use a mapping dictionary and a loop to build the complement.

````python
def complement(sequence):
    mapping = {"A": "T", "T": "A", "G": "C", "C": "G"}
    result = ""

    for base in sequence:
        result += mapping[base]

    return result

print(complement("ATGC"))
````

## 40. Integrated Bioinformatics Example

````python
sequences = {
    "Gene_A": "ATGCGTACG",
    "Gene_B": "GGCCGCGC",
    "Gene_C": "ATATATAT",
    "Gene_D": "GCGCGCGC"
}

def gc_content(sequence):
    gc = sequence.count("G") + sequence.count("C")
    return (gc / len(sequence)) * 100

for gene, sequence in sequences.items():
    print("Gene:", gene)
    print("Length:", len(sequence))
    print("GC content:", round(gc_content(sequence), 2), "%")
    print()
````

## 41. Recursion

A recursive function calls itself.

A recursive function needs:

1. a **base case** — when to stop
2. a **recursive case** — the function calls itself with a smaller/simpler problem

````python
def countdown(n):
    if n == 0:
        return

    print(n)
    countdown(n - 1)

countdown(5)
````

## 42. Factorial Using Recursion

````python
def factorial(n):
    if n == 0:
        return 1

    return n * factorial(n - 1)

print(factorial(5))
````

## 43. Recursion vs Iteration

````python
def countdown_loop(n):
    while n > 0:
        print(n)
        n -= 1

countdown_loop(5)
````

````python
def countdown_recursive(n):
    if n == 0:
        return

    print(n)
    countdown_recursive(n - 1)

countdown_recursive(5)
````

Both approaches can express repeated computation.

The key difference is that a loop repeats through iteration, while recursion solves a problem by calling the same function again.
