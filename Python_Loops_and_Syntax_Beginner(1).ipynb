# 3. The `for` Loop

A `for` loop is used when we want to repeat something for each item in a sequence.

### Basic syntax

```python
for variable in sequence:
    statement
```

The indented code runs once for every item.


````python
for i in [1, 2, 3, 4, 5]:
    print(i)
````

# 9. The `while` Loop

A `while` loop repeats as long as a condition is `True`.

### Basic syntax

```python
while condition:
    statement
```

Always make sure the condition can eventually become `False`; otherwise, the loop may run forever.


````python
count = 1

while count <= 5:
    print(count)
    count = count + 1
````

## 10. `for` vs `while`

| `for` loop | `while` loop |
|---|---|
| Useful for sequences and known repetitions | Useful when repetition depends on a condition |
| Often used with `range()` | Requires a condition |
| Number of iterations is often predictable | Number of iterations may not be known in advance |

Example:

```python
for i in range(5):
    print(i)
```

versus

```python
i = 0
while i < 5:
    print(i)
    i += 1
```


````python
# for loop

for i in range(5):
    print(i)
````

````python
# while loop

i = 0

while i < 5:
    print(i)
    i += 1
````

## 11. The `break` Statement

`break` immediately stops a loop.


````python
for i in range(1, 11):
    if i == 6:
        break
    print(i)
````

## 12. The `continue` Statement

`continue` skips the current iteration and moves to the next one.


````python
for i in range(1, 11):
    if i == 5:
        continue
    print(i)
````

## 13. Nested Loops

A loop inside another loop is called a **nested loop**.

The inner loop runs completely for every iteration of the outer loop.


````python
for i in range(1, 4):
    for j in range(1, 4):
        print("i =", i, "j =", j)
````

### Example: Simple Multiplication Table

````python
for i in range(1, 6):
    for j in range(1, 6):
        print(i * j, end=" ")
    print()
````

# 16. A Small Biological Programming Exercise

Calculate the **GC content** of a DNA sequence.

GC content is:

$$
GC\% = \frac{G + C}{\text{Total number of bases}} \times 100
$$


````python
sequence = "ATGCGCGTATGC"

gc_count = 0

for base in sequence:
    if base == "G" or base == "C":
        gc_count += 1

gc_content = (gc_count / len(sequence)) * 100

print("GC count:", gc_count)
print("GC content:", gc_content, "%")
````

# 17. Common Syntax Errors

### Error 1: Missing colon

Incorrect:

```python
for i in range(5)
    print(i)
```

Correct:

```python
for i in range(5):
    print(i)
```

### Error 2: Incorrect indentation

Incorrect:

```python
for i in range(5):
print(i)
```

Correct:

```python
for i in range(5):
    print(i)
```

### Error 3: Infinite `while` loop

Be careful with:

```python
i = 1

while i <= 5:
    print(i)
```

The value of `i` never changes, so the condition never becomes false.
